```python
from google.colab import userdata
import os
import subprocess

GITHUB_USER = 'NemikaA'
REPO_OWNER = 'NemikaA'
REPO_NAME = 'Neuro-X'
BRANCH_NAME = 'feature/mri-preprocessing'
REPO_DIR = f'/content/{REPO_NAME}'

token = userdata.get('GITHUB_TOKEN')
repo_url = f'https://{GITHUB_USER}:{token}@github.com/{REPO_OWNER}/{REPO_NAME}.git'

if not os.path.exists(REPO_DIR):
    subprocess.run(['git', 'clone', '-b', BRANCH_NAME, repo_url, REPO_DIR], check=True)
else:
    subprocess.run(['git', '-C', REPO_DIR, 'pull', 'origin', BRANCH_NAME], check=True)

os.chdir(REPO_DIR)
print('Working directory:', os.getcwd())
```

```python
import os
import subprocess

USE_FULL_DATASET = False
ZIP_NAME = 'samplee.zip' if USE_FULL_DATASET else 'sampleee_dataset.zip'
ZIP_PATH = os.path.join(DATA_DIR, ZIP_NAME)
EXTRACT_DIR = '/content/local_dataset'

os.makedirs(EXTRACT_DIR, exist_ok=True)
subprocess.run(['df', '-h', '/content'], check=True)
subprocess.run(['unzip', '-q', '-o', ZIP_PATH, '-d', EXTRACT_DIR], check=True)

print('Extracted contents:', os.listdir(EXTRACT_DIR))
```
